```python
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import os
import re
import numpy as np
```

```python
adata = sc.read_h5ad("../preprocessed/step1_umap_annotate/Reyes_multiome_BATCH_m1_SUBSET_K2ctrl__embed.h5ad")
```

```python
mean_counts = np.mean(np.ravel(adata.layers['cellbender'].sum(axis = 1)))
print("Dataset_id = " + np.unique(adata.obs['prefix'])[0])
print("Mean counts = " + str(mean_counts))
print("Total cells = " + str(adata.n_obs))
print("Total pcs = " + str(adata.obsm['X_pca'].shape[1]))
print("Explained var = " + str(np.sum(adata.uns['pca']['variance_ratio'])))
```

```python
adata.obs['cell_type'].value_counts() / adata.n_obs
```
